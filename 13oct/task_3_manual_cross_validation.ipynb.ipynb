## Use cross-validation to ensure model generalization by manually splitting data into k-folds, training and testing models iteratively without using built-in cross-validation functions, and computing aggregated performance metrics.

```python
import pandas as pd
import random
from math import sqrt
from collections import Counter
```

```python
df = pd.read_csv("heart.csv")

X = df.drop("target", axis=1).values.tolist()
y = df["target"].tolist()
```

```python

random.seed(42)

data = list(zip(X, y))
random.shuffle(data)

k = 5
fold_size = len(data) // k
folds = []

for i in range(k):
    start = i * fold_size
    end = start + fold_size

    if i == k - 1:
        folds.append(data[start:])
    else:
        folds.append(data[start:end])

print("Total folds:", len(folds))
```

```python

scores = []

for i in range(k):
    test_data = folds[i]
    train_data = []

    for j in range(k):
        if j != i:
            train_data += folds[j]

    train_labels = [row[1] for row in train_data]
    test_labels = [row[1] for row in test_data]

    prediction = Counter(train_labels).most_common(1)[0][0]
    predictions = [prediction] * len(test_labels)

    correct = sum(a == b for a, b in zip(test_labels, predictions))
    accuracy = correct / len(test_labels) * 100

    scores.append(accuracy)
    print("Fold", i + 1, "Accuracy:", round(accuracy, 2), "%")
```

```python

mean_accuracy = sum(scores) / len(scores)

variance = sum((s - mean_accuracy) ** 2 for s in scores) / len(scores)
std_deviation = sqrt(variance)

print("Average Accuracy:", round(mean_accuracy, 2), "%")
print("Accuracy Standard Deviation:", round(std_deviation, 2))
```
