### Load the dataset in a Streamlit app without using built-in Pandas functions, manually parse the file format and delimiter, and display the dataset with pagination and sorting options.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
# Pagination
page = 1
rows = 5

start = (page - 1) * rows
end = start + rows

print(df.iloc[start:end])
```

```python
# Sorting
df = df.sort_values("age")

print(df)
```
