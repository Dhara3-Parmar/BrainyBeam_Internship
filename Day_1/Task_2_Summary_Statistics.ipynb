### Show summary statistics (mean, median, mode, min, max, std) by implementing manual calculations without using built-in statistical functions.

```python
import pandas as pd

df = pd.read_csv("heart.csv")

df.head()
```

```python
values = list(df["age"])

print(values[:10])
```

```python
total = 0

for x in values:
    total = total + x

mean = total / len(values)

print("Mean =", mean)
```

```python
values.sort()

n = len(values)

if n % 2 == 0:
    median = (values[n//2 - 1] + values[n//2]) / 2
else:
    median = values[n//2]

print("Median =", median)
```

```python
mode = values[0]
max_count = 0

for x in values:
    count = values.count(x)

    if count > max_count:
        max_count = count
        mode = x

print("Mode =", mode)
```

```python
minimum = values[0]
maximum = values[0]

for x in values:
    if x < minimum:
        minimum = x

    if x > maximum:
        maximum = x

print("Minimum =", minimum)
print("Maximum =", maximum)
```

```python
total = 0

for x in values:
    total = total + (x - mean) ** 2

std = (total / len(values)) ** 0.5

print("Standard Deviation =", std)
```

```python
print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
print("Minimum =", minimum)
print("Maximum =", maximum)
print("Standard Deviation =", std)
```
