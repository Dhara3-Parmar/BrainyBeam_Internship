### Identify missing values without using .isna() or .dropna(), apply multiple imputation techniques, and justify the method chosen in a report.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("heart.csv")
df.head()
```

```python
for column in df.columns:
    count = 0

    for value in df[column]:
        if pd.isna(value):
            count += 1

    print(column, ":", count)
```

```python
for column in df.columns:
    count = 0

    for value in df[column]:
        if pd.isna(value):
            count += 1

    if count > 0:
        print(column, "has", count, "missing values")
```

```python
column = "age"

values = []

for value in df[column]:
    if not pd.isna(value):
        values.append(value)

mean = sum(values) / len(values)

print("Mean:", mean)
```

```python
values.sort()

middle = len(values) // 2

if len(values) % 2 == 0:
    median = (values[middle-1] + values[middle]) / 2
else:
    median = values[middle]

print("Median:", median)
```

```python
column = "sex"

values = []

for value in df[column]:
    if not pd.isna(value):
        values.append(value)

mode = max(set(values), key=values.count)

print("Mode:", mode)
```

```python
for column in df.columns:
    if df[column].dtype != "object":

        values = []

        for value in df[column]:
            if not pd.isna(value):
                values.append(value)

        values.sort()
        middle = len(values) // 2

        if len(values) % 2 == 0:
            median = (values[middle-1] + values[middle]) / 2
        else:
            median = values[middle]

        for i in range(len(df[column])):
            if pd.isna(df.loc[i, column]):
                df.loc[i, column] = median
```

```python
for column in df.columns:
    if df[column].dtype == "object":

        values = []

        for value in df[column]:
            if not pd.isna(value):
                values.append(value)

        if len(values) > 0:
            mode = max(set(values), key=values.count)

            for i in range(len(df[column])):
                if pd.isna(df.loc[i, column]):
                    df.loc[i, column] = mode
```

```python
for column in df.columns:
    count = 0

    for value in df[column]:
        if pd.isna(value):
            count += 1

    print(column, ":", count)
```
