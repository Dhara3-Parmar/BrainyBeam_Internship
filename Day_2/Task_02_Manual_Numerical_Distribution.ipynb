### Visualize the distribution of all numerical columns using only raw Matplotlib commands, ensuring custom aesthetics, annotations, and multiple comparative plots for deeper insights.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
columns = df.select_dtypes(include="number").columns
print(columns)
```

```python
for column in columns:

    data = df[column].dropna()

    plt.figure(figsize=(6, 4))
    plt.hist(data, bins=10, edgecolor="black")

    plt.title("Distribution of " + column)
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.axvline(data.mean(), linestyle="--")
    plt.text(data.mean(), 5, "Mean")

    plt.show()
```

```python
means = []

for column in columns:
    means.append(df[column].mean())

plt.figure(figsize=(8, 4))
plt.bar(columns, means, edgecolor="black")

plt.title("Mean Comparison of Numerical Features")
plt.xlabel("Features")
plt.ylabel("Mean")

plt.xticks(rotation=45)

for i in range(len(columns)):
    plt.text(i, means[i], round(means[i], 1), ha="center")

plt.show()
```

```python
medians = []

for column in columns:
    medians.append(df[column].median())

plt.figure(figsize=(8, 4))
plt.bar(columns, medians, edgecolor="black")

plt.title("Median Comparison of Numerical Features")
plt.xlabel("Features")
plt.ylabel("Median")

plt.xticks(rotation=45)

for i in range(len(columns)):
    plt.text(i, medians[i], round(medians[i], 1), ha="center")

plt.show()
```
