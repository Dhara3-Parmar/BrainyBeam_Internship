### Convert categorical columns (sex, cp, thal, etc.) into numerical form without using built-in encoding functions, manually mapping categories and handling unknown values dynamically.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
categorical = df.select_dtypes(include="object").columns

print(categorical)
```

```python
def manual_encode(data):
    
    mapping = {}
    result = []
    
    for value in data:
        
        if value not in mapping:
            mapping[value] = len(mapping)
        
        result.append(mapping[value])
    
    return result, mapping
```

```python
for column in categorical:
    
    df[column], mapping = manual_encode(df[column])
    
    print(column, ":", mapping)
```
