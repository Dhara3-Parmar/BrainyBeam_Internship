```python
numerical_columns = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

print(numerical_columns)
```

```python
#manually identify outliers
for column in numerical_columns:

    values = sorted(df[column].tolist())

    n = len(values)

    q1_index = int(0.25 * (n - 1))
    q3_index = int(0.75 * (n - 1))

    q1 = values[q1_index]
    q3 = values[q3_index]

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outlier_count = 0

    for value in values:
        if value < lower_limit or value > upper_limit:
            outlier_count += 1

    print("\nColumn:", column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", outlier_count)
```

```python
for column in numerical_columns:

    values = sorted(df[column].tolist())

    n = len(values)

    q1_index = int(0.25 * (n - 1))
    q3_index = int(0.75 * (n - 1))

    q1 = values[q1_index]
    q3 = values[q3_index]

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outlier_count = 0

    for value in values:
        if value < lower_limit or value > upper_limit:
            outlier_count += 1

    print("\nColumn:", column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", outlier_count)
```

```python
#mean and standard division
df_clean = df.copy()

# Convert numerical columns to float
for column in numerical_columns:
    df_clean[column] = df_clean[column].astype(float)


for column in numerical_columns:

    values = sorted(df_clean[column].tolist())

    n = len(values)

    q1_index = int(0.25 * (n - 1))
    q3_index = int(0.75 * (n - 1))

    q1 = values[q1_index]
    q3 = values[q3_index]

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    for index in df_clean.index:

        value = df_clean.loc[index, column]

        if value < lower_limit:
            df_clean.loc[index, column] = lower_limit

        elif value > upper_limit:
            df_clean.loc[index, column] = upper_limit


print("Outliers handled successfully.")
```

```python
for column in numerical_columns:

    values = df_clean[column].tolist()

    total = 0

    for value in values:
        total = total + value

    mean = total / len(values)

    squared_sum = 0

    for value in values:
        squared_sum = squared_sum + (value - mean) ** 2

    variance = squared_sum / len(values)

    std = variance ** 0.5

    print("\n", column)
    print("Mean:", mean)
    print("Standard Deviation:", std)
```

```python
df_standardized = df_clean.copy()

for column in numerical_columns:

    values = df_clean[column].tolist()

    total = 0

    for value in values:
        total += value

    mean = total / len(values)

    squared_sum = 0

    for value in values:
        squared_sum += (value - mean) ** 2

    variance = squared_sum / len(values)

    std = variance ** 0.5

    if std != 0:

        for index in df_standardized.index:

            value = df_standardized.loc[index, column]

            df_standardized.loc[index, column] = (
                value - mean
            ) / std

print("Manual standardization completed.")
```
