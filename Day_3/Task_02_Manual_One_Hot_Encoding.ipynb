### Perform one-hot encoding manually by implementing a custom encoding function that dynamically adapts to new unseen categories in future data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
categorical_columns = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

print("Categorical columns:")
print(categorical_columns)
```

```python
def manual_one_hot_encode(data, columns):

    result = data.copy()

    for column in columns:

        categories = []

        for value in result[column]:

            if value not in categories:
                categories.append(value)

        print("\nColumn:", column)
        print("Categories:", categories)

        for category in categories:

            new_column = column + "_" + str(category)

            result[new_column] = 0

            for index in result.index:

                if result.loc[index, column] == category:
                    result.loc[index, new_column] = 1

        result.drop(column, axis=1, inplace=True)

    return result
```

```python
df_encoded = manual_one_hot_encode(
    df,
    categorical_columns
)
```
