## Create a new feature based on chol and thalach using a complex formula incorporating domain knowledge, interaction terms, and percentile-based ranking.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
def manual_percentile(values):

    sorted_values = sorted(values)

    result = []

    total = len(sorted_values)

    for value in values:

        count = 0

        for item in sorted_values:

            if item <= value:
                count += 1

        percentile = (count / total) * 100

        result.append(percentile)

    return result
```

```python
df["chol_percentile"] = manual_percentile(
    df["chol"].tolist()
)
```

```python
df["thalach_percentile"] = manual_percentile(
    df["thalach"].tolist()
)
```

```python
display(
    df[
        [
            "chol",
            "chol_percentile",
            "thalach",
            "thalach_percentile"
        ]
    ].head(10)
)
```

```python
df["chol_risk"] = (
    df["chol_percentile"] / 100
)

df["thalach_risk"] = (
    1 - (df["thalach_percentile"] / 100)
)
```

```python
df["chol_thalach_interaction"] = (
    df["chol_risk"] * df["thalach_risk"]
)
```

```python
df["risk_factor"] = (
    (0.50 * df["chol_risk"])
    +
    (0.30 * df["thalach_risk"])
    +
    (0.20 * df["chol_thalach_interaction"])
)
```

```python
display(
    df[
        [
            "chol",
            "thalach",
            "chol_percentile",
            "thalach_percentile",
            "chol_risk",
            "thalach_risk",
            "chol_thalach_interaction",
            "risk_factor"
        ]
    ].head(20)
)
```

```python
result = df[
    [
        "age",
        "chol",
        "thalach",
        "risk_factor",
        "target"
    ]
].sort_values(
    "risk_factor",
    ascending=False
)

display(result.head(10))
```
