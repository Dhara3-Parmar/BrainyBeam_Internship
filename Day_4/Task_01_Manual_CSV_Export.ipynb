### Store processed data in a new CSV file without using built-in DataFrame export functions, manually formatting and writing the data while ensuring optimal storage efficiency.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
file = open("heart_processed.csv", "w")

file.write(",".join(df.columns) + "\n")

for row in df.values:
    file.write(",".join(map(str, row)) + "\n")

file.close()

print("File saved successfully")
```

```python
import os

print("File exists:", os.path.exists("heart_processed.csv"))
print("File size:", os.path.getsize("heart_processed.csv"), "bytes")
```
