### Create histograms for all numerical features without using built-in plotting functions, implementing binning and frequency calculations manually.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
numerical_columns = df.select_dtypes(include="number").columns

print(numerical_columns)
```

```python
for column in numerical_columns:
    
    data = df[column].dropna().tolist()
    
    minimum = min(data)
    maximum = max(data)
    bins = 5
    width = (maximum - minimum) / bins
    frequency = [0] * bins
    
    for value in data:
        index = int((value - minimum) / width)
        if index == bins:
            index = bins - 1
        frequency[index] += 1
    
    print("\n", column)
    print("Frequency:", frequency)
```

```python
for column in numerical_columns:
    
    data = df[column].dropna().tolist()
    minimum = min(data)
    maximum = max(data)
    bins = 5
    width = (maximum - minimum) / bins
    frequency = [0] * bins
    
    for value in data:
        index = int((value - minimum) / width)
        if index == bins:
            index = bins - 1
        frequency[index] += 1
    
    x = range(bins)
    
    plt.figure(figsize=(5, 3))
    plt.bar(x, frequency)
    plt.title(column)
    plt.xlabel("Bins")
    plt.ylabel("Frequency")
    plt.show()
```
