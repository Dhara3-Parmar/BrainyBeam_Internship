### Display bar plots for categorical variables without using high-level plotting libraries, dynamically adjusting bar widths and colors for better interpretation.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
categorical = ["sex", "cp", "fbs", "restecg",
               "exang", "slope", "ca", "thal"]

print(categorical)
```

```python
column = "sex"

print(df[column].unique())
```

```python
categories = []
counts = []

for value in df[column]:
    if value not in categories:
        categories.append(value)
        counts.append(1)
    else:
        i = categories.index(value)
        counts[i] += 1

print(categories)
print(counts)
```

```python
width = 0.8

if len(categories) > 5:
    width = 0.5
print("Bar width:", width)
```

```python
colors = []

for i in range(len(categories)):
    colors.append("C" + str(i))
print(colors)
```

```python
plt.bar(categories, counts, width=width, color=colors)

plt.title("Distribution of " + column)
plt.xlabel(column)
plt.ylabel("Frequency")

plt.show()
```

```python
for column in categorical:

    categories = []
    counts = []

    for value in df[column]:
        if value not in categories:
            categories.append(value)
            counts.append(1)
        else:
            i = categories.index(value)
            counts[i] += 1

    width = 0.8

    if len(categories) > 5:
        width = 0.5

    colors = []

    for i in range(len(categories)):
        colors.append("C" + str(i))

    plt.figure(figsize=(6, 4))
    plt.bar(categories, counts, width=width, color=colors)

    plt.title("Distribution of " + column)
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()
```
