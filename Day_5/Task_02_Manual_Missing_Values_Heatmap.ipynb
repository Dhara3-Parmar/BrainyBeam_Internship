### Create a heatmap of missing values (if applicable) using a manually constructed grid, mapping missing data points with custom color gradients instead of using Seaborn or Matplotlib heatmap functions.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
missing = []

for column in df.columns:
    count = 0
    
    for value in df[column]:
        if pd.isnull(value):
            count += 1
    
    missing.append(count)

print(missing)
```

```python
grid = []

for row in range(len(df)):
    line = []
    
    for column in df.columns:
        if pd.isnull(df.loc[row, column]):
            line.append(1)
        else:
            line.append(0)
    
    grid.append(line)
```

```python
plt.figure(figsize=(12, 6))

plt.imshow(grid, aspect="auto", cmap="Greys")

plt.xticks(range(len(df.columns)), df.columns, rotation=45)
plt.xlabel("Columns")
plt.ylabel("Rows")

plt.title("Manual Missing Values Heatmap")

plt.show()
```

```python
total_missing = 0

for value in missing:
    total_missing += value

if total_missing == 0:
    print("No missing values found in the dataset.")
else:
    print("Total missing values:", total_missing)
```
