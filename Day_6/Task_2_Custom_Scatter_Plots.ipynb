### Use scatter plots to analyze relationships (e.g., chol vs thalach) by implementing a custom plotting function that dynamically adjusts point sizes and colors based on additional attributes like age or target.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
print(df.columns)
```

```python
print(df[["chol", "thalach", "age", "target"]].isnull().sum())
```

```python
def custom_scatter(x, y, size_data, color_data, x_name, y_name):

    sizes = []

    for value in size_data:
        sizes.append(value * 3)

    plt.figure(figsize=(8, 5))

    plt.scatter(
        x,
        y,
        s=sizes,
        c=color_data,
        cmap="coolwarm",
        alpha=0.7
    )

    plt.xlabel(x_name)
    plt.ylabel(y_name)

    plt.title(x_name + " vs " + y_name)

    plt.colorbar(label="Target")

    plt.show()
```

```python
def custom_scatter(x, y, size_data, color_data, x_name, y_name):

    sizes = []

    for value in size_data:
        s = value * 3
        sizes.append(s)

    plt.figure(figsize=(8, 5))

    plt.scatter(
        x,
        y,
        s=sizes,
        c=color_data,
        cmap="coolwarm",
        alpha=0.7
    )

    plt.xlabel(x_name)
    plt.ylabel(y_name)

    plt.title(x_name + " vs " + y_name)

    plt.colorbar(label="Target")

    plt.show()
```

```python
custom_scatter(
    df["chol"],
    df["thalach"],
    df["age"],
    df["target"],
    "Cholesterol",
    "Maximum Heart Rate"
)
```

```python
custom_scatter(
    df["age"],
    df["thalach"],
    df["chol"],
    df["target"],
    "Age",
    "Maximum Heart Rate"
)
```

```python
print(df["target"].value_counts())
```

```python
target_0 = df[df["target"] == 0]

target_1 = df[df["target"] == 1]

print("Target 0:", len(target_0))
print("Target 1:", len(target_1))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    target_0["chol"],
    target_0["thalach"],
    s=target_0["age"] * 3,
    alpha=0.6,
    label="Target 0"
)

plt.scatter(
    target_1["chol"],
    target_1["thalach"],
    s=target_1["age"] * 3,
    alpha=0.6,
    label="Target 1"
)

plt.xlabel("Cholesterol")
plt.ylabel("Maximum Heart Rate")

plt.title("Cholesterol vs Maximum Heart Rate")

plt.legend()

plt.show()
```
