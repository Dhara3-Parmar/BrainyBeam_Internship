### Perform a chi-square test for categorical columns and target by manually computing observed and expected frequencies, applying the chi-square formula step by step, and validating results without using statistical libraries.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
categorical = ["sex", "cp", "fbs", "restecg",
               "exang", "slope", "ca", "thal"]

print(categorical)
```

```python
target = "target"
```

```python
col = "sex"

observed = {}

for value in df[col].unique():
    observed[value] = {}

    for t in df[target].unique():
        count = 0

        for i in range(len(df)):
            if df[col].iloc[i] == value and df[target].iloc[i] == t:
                count += 1

        observed[value][t] = count

print(observed)
```

```python
row_total = {}

for value in observed:
    row_total[value] = sum(observed[value].values())

print(row_total)
```

```python
col_total = {}

for t in df[target].unique():
    col_total[t] = sum(observed[v][t] for v in observed)

print(col_total)
```

```python
total = sum(col_total.values())

print("Total:", total)
```

```python
expected = {}

for value in observed:
    expected[value] = {}

    for t in df[target].unique():
        expected[value][t] = (
            row_total[value] * col_total[t] / total
        )

print(expected)
```

```python
chi_square = 0

for value in observed:
    for t in df[target].unique():

        O = observed[value][t]
        E = expected[value][t]

        chi_square += (O - E) ** 2 / E

print("Chi-Square Value:", round(chi_square, 3))
```

```python
rows = len(observed)
columns = len(df[target].unique())

df_value = (rows - 1) * (columns - 1)

print("Degrees of Freedom:", df_value)
```

```python
critical_values = {
    1: 3.841,
    2: 5.991,
    3: 7.815,
    4: 9.488,
    5: 11.070,
    6: 12.592,
    7: 14.067,
    8: 15.507,
    9: 16.919,
    10: 18.307
}
```

```python
critical = critical_values[df_value]

print("Chi-Square:", round(chi_square, 3))
print("Critical Value:", critical)

if chi_square > critical:
    print("Significant relationship with target")
else:
    print("No significant relationship with target")
```

```python
for col in categorical:

    values = df[col].unique()
    targets = df[target].unique()

    observed = {}

    for v in values:
        observed[v] = {}

        for t in targets:
            count = 0

            for i in range(len(df)):
                if df[col].iloc[i] == v and df[target].iloc[i] == t:
                    count += 1

            observed[v][t] = count

    row_total = {}
    for v in values:
        row_total[v] = sum(observed[v].values())

    col_total = {}
    for t in targets:
        col_total[t] = sum(observed[v][t] for v in values)

    total = sum(col_total.values())

    chi = 0

    for v in values:
        for t in targets:

            O = observed[v][t]
            E = row_total[v] * col_total[t] / total

            if E != 0:
                chi += (O - E) ** 2 / E

    df_value = (len(values) - 1) * (len(targets) - 1)

    print(col, "=", round(chi, 3))
```
